## 2 · The LCA boundary matters

**Foreground: gate-to-gate, operation only. Background: associated supply and treatment burdens remain included through the mapped datasets.**

The modeled foreground is the continuous production train during production and modeled cleaning-in-place (CIP), with its specified inventory channels. It is not a complete product life cycle or a separate assessment of plant construction and capital equipment; background datasets retain their own included burdens. The functional unit in the manuscript is **40 mg of purified lovastatin API**; this normalization does not include downstream distribution, product use or end-of-life.

Electricity-generation shares for DK2 vary hourly. Other background datasets remain static. Do not confuse a static background dataset with a necessarily constant process demand.

**Important CIP simplification:** the manuscript uses a wastewater-only proxy and excludes cleaning electricity and cleaning-agent inventories. The corresponding activity/unit mapping in the original notebook still requires checking; the presentation does not resolve it.

## 4 · Separate the process model from the LCA contribution

**KTB1 is an external dynamic process model, developed by Boskabadi and collaborators.** It provides the simulation side of the example. The manuscript describes it as a digital twin; this repository's supplied evidence is simulation-based, not a verified connection to a physical plant.

[Open the KTB1 repository](https://github.com/Boskabadi/KTB1-DLCA) · [Read the model paper](https://doi.org/10.1016/j.compchemeng.2024.108770)

Model credit: Boskabadi, M. R.; Ramin, P.; Kager, J.; Sin, G.; Mansouri, S. S. *KT-Biologics I (KTB1): A Dynamic Simulation Model for Continuous Biologics Manufacturing.* Computers & Chemical Engineering **188** (2024), 108770.

The supplied main notebook expects `KTB1_DLCA1_newvariable_22_legacy` and a `step_and_get_DI` helper. Those exact dependencies are not present in the inspected upstream snapshot. Obtain the correct model revision and helper before attempting execution; no substitute model has been added here.


## 5 · Understand the campaign calendar

The case-study assumption is a **480-hour production block followed by 120 hours of CIP**. Nine such 600-hour cycles make 5,400 operating hours within an annual planning window. Scheduling compares where those cycles start; it is not a demonstration of physical actuation.

This saved two-cycle plot illustrates the block structure only. It does not validate the chosen industrial operating or cleaning durations.
