# 01 · MATLAB–Python connection

The framework demonstration obtains process data from a MATLAB/Simulink model.
This notebook documents an optional local setup; **Run All installs nothing and
starts no MATLAB session**.

The supplied model metadata identifies MATLAB **R2025b (25.2)**. The historical
analysis notebook instead listed `matlabengine 25.1`; that is an unresolved
environment-provenance inconsistency, not a tested pairing.

## Check your own installation

1. Use your institutionally licensed MATLAB/Simulink installation.
2. Select a Python environment compatible with that MATLAB release.
3. Install the matching engine into that same environment, then select its
   kernel in Jupyter. None of these steps have been executed for this package.

The official MathWorks R2025b engine README lists Python 3.9–3.12 and gives the
following installation example. Run it manually only after confirming that
R2025b is your installed release:

```bash
python -m pip install matlabengine==25.2.2
```

For another MATLAB release, follow its matching MathWorks instructions rather
than installing an unpinned latest engine. Do not uninstall unrelated packages
or request administrator rights automatically.
